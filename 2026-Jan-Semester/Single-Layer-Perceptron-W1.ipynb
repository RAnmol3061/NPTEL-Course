## Problem Statement

Inputs: x = [x_1, x_2] = [1.0, 2.0]

Target: y = 1

Weights: w = [w_1, w_2] = [0.2, -0.5]

Bias: b = 0.1

Learning rate n = 0.1

Prediction Rule if computed value is less than or equal to 0 then 0 otherwise 1

```python
import numpy as np
```

```python
# Define initial values and parameters

x = np.array([1.0, 2.0])  # Input
y = 1.0                   # Target
w = np.array([0.2, -0.5])  # Weights
b = 0.1                   # Bias
eta = 0.1                 # Learning rate
epoch = 1                 # Counter
```

```python
while True:

    print(f"\nCurrent epoch: {epoch}")

    computed_value = np.dot(w,x) + b
    y_pred = 1 if computed_value > 0 else 0
    error = y - y_pred

    print(f"Computed Value: {computed_value:.2f}")
    print(f"Prediction: {y_pred}")
    print(f"Error: {error}")


    #Update Rule
    w +=  error * eta * x
    b +=  error * eta

    print(f"Updated Weights: {w}")
    print(f"Updated Bias: {b:.2f}")

    if error == 0 or epoch > 10: #OR
        break   
    else:
        epoch += 1
```

```python
#Implementation

x_test = np.array([2.0,1.0])
b_test = 0.4

computed_value_test = np.dot(w,x_test) + b_test
y_pred_test = 1 if computed_value_test > 0 else 0

print(f"Test Input: {x_test}")
print(f"Prediction: {y_pred_test}")
```
